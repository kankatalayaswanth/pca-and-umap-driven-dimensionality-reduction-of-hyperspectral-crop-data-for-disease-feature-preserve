# Notebook 3: Dimensionality Reduction (PCA & UMAP)
**Course:** INT396 - Unsupervised Learning

This notebook implements and compares:
1. Linear Global Projection: Principal Component Analysis (PCA)
2. Non-linear Manifold Learning: Uniform Manifold Approximation and Projection (UMAP)
3. Scree plots, cumulative explained variance, and component loadings

```python
import sys
sys.path.append('..')

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from src.data_loader import load_dataset
from src.preprocessing import build_preprocessing_pipeline, fit_transform_pipeline
from src.pca import fit_pca, compute_scree_data, compute_loadings, get_top_contributing_bands, transform_pca
from src.umap_reduction import fit_umap

X, y, meta, bands, wls = load_dataset('potato_dai70_vnir')
pipe = build_preprocessing_pipeline('standard')
X_sc, _, act_bands = fit_transform_pipeline(pipe, X, band_names=bands)
print(f"Standardized feature matrix: {X_sc.shape}")
```

```python
# 1. PCA Analysis
pca, k = fit_pca(X_sc, n_components=5)
X_pca = transform_pca(pca, X_sc)
scree = compute_scree_data(pca)
print("Scree table:
", scree)
```

```python
# 2. UMAP Embedding
reducer, X_umap = fit_umap(X_sc, n_components=2, n_neighbors=10, min_dist=0.1, random_state=42)
print(f"UMAP embedding shape: {X_umap.shape}")
```

```python
# 3. Visual Comparison of PCA 2D vs UMAP 2D
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5), facecolor='white')

# PCA plot
df_pca = pd.DataFrame(X_pca[:, :2], columns=['PC1', 'PC2'])
df_pca['Disease'] = y.values
sns.scatterplot(data=df_pca, x='PC1', y='PC2', hue='Disease', palette='tab10', ax=ax1, s=60)
ax1.set_title("PCA 2D Projection (PC1 vs PC2)", fontweight='bold')
ax1.grid(True, linestyle='--', alpha=0.3)

# UMAP plot
df_umap = pd.DataFrame(X_umap, columns=['UMAP1', 'UMAP2'])
df_umap['Disease'] = y.values
sns.scatterplot(data=df_umap, x='UMAP1', y='UMAP2', hue='Disease', palette='tab10', ax=ax2, s=60)
ax2.set_title("UMAP 2D Manifold Embedding", fontweight='bold')
ax2.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```
