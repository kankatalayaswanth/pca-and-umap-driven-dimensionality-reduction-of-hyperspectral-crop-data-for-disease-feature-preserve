# Notebook 2: Preprocessing Pipeline & Leakage-Free Splitting
**Course:** INT396 - Unsupervised Learning

This notebook demonstrates the construction of a Scikit-Learn preprocessing pipeline:
- SimpleImputer for robust missing value handling
- VarianceThreshold for constant feature elimination
- StandardScaler / MinMaxScaler
- Leakage-free train/validation split evaluation

```python
import sys
sys.path.append('..')

import numpy as np
import pandas as pd
from src.data_loader import load_dataset
from src.preprocessing import (
    build_preprocessing_pipeline,
    create_train_test_split,
    fit_transform_pipeline,
    compare_before_after
)

X, y, meta, bands, wls = load_dataset('potato_dai70_vnir')
print(f"Original shape: {X.shape}")
```

```python
# 1. Split into train and test sets to guarantee no data leakage
split = create_train_test_split(X, y, test_size=0.25, random_state=42)
X_train, X_test = split['X_train'], split['X_test']
print(f"Train samples: {len(X_train)}, Test samples: {len(X_test)}")
```

```python
# 2. Build and fit pipeline strictly on training data
pipeline = build_preprocessing_pipeline(scaler_type='standard', variance_threshold=0.0)
X_train_proc, X_test_proc, act_bands = fit_transform_pipeline(pipeline, X_train, X_test, band_names=bands)

print(f"X_train_proc mean: {X_train_proc.mean():.4f}, std: {X_train_proc.std():.4f}")
print(f"X_test_proc shape: {X_test_proc.shape}")
```

```python
# 3. Before vs After summary
summary = compare_before_after(X_train, X_train_proc, act_bands)
print("First 5 bands transformation summary:")
summary.head(5)
```
